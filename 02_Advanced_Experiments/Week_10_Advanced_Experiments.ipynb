## Experiment 1: Standardize two features

**Purpose.** Scaling prevents larger-unit attributes from dominating Euclidean clustering.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import numpy as np
from sklearn.preprocessing import StandardScaler
X=np.array([[1.,100.],[2.,200.],[3.,400.]])
Z=StandardScaler().fit_transform(X)
print(np.round(Z,2))
```

### How to interpret this experiment

Scaling prevents larger-unit attributes from dominating Euclidean clustering.

- `StandardScaler` — Centers and scales columns using values learned from training data.

## Experiment 2: Fit K-means reproducibly

**Purpose.** K-means labels are arbitrary identifiers, not ordered scientific categories.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import numpy as np
from sklearn.cluster import KMeans
X=np.array([[0,0],[0,1],[8,8],[8,9]],dtype=float)
labels=KMeans(n_clusters=2,random_state=2026,n_init=10).fit_predict(X)
print(labels)
```

### How to interpret this experiment

K-means labels are arbitrary identifiers, not ordered scientific categories.

- `KMeans` — Splits feature vectors into a specified number of compact clusters.

**Questions:** (1) What does the printed output measure? (2) What happens if one input or parameter changes? (3) How could this method be extended to an independent geographic dataset?

## Experiment 3: DBSCAN and noise labels

**Purpose.** DBSCAN noise label -1 means not assigned to a density-connected cluster.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import numpy as np
from sklearn.cluster import DBSCAN
XY=np.array([[0,0],[0,1],[1,0],[10,10]],dtype=float)
labels=DBSCAN(eps=1.5,min_samples=3).fit_predict(XY)
print(labels)
```

### How to interpret this experiment

DBSCAN noise label -1 means not assigned to a density-connected cluster.

- `DBSCAN` — Clusters observations by local density under eps and min_samples.

**Questions:** (1) What does the printed output measure? (2) What happens if one input or parameter changes? (3) How could this method be extended to an independent geographic dataset?

## Experiment 4: Silhouette validity check

**Purpose.** Silhouette is a descriptive cluster-separation index, not a proof of real-world classes.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import numpy as np
from sklearn.metrics import silhouette_score
X=np.array([[0,0],[0,1],[8,8],[8,9]],dtype=float)
labels=np.array([0,0,1,1])
print('Silhouette:',silhouette_score(X,labels))
```

### How to interpret this experiment

Silhouette is a descriptive cluster-separation index, not a proof of real-world classes.
