## Experiment 1: Create geographic block groups

**Purpose.** Blocks define what constitutes a new spatial group.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import pandas as pd
cells=pd.read_csv(DATA/'cells.csv')
cells['block_id']=(cells['row']//3).astype(str)+'_'+(cells['col']//3).astype(str)
print(cells.groupby('block_id').size().head())
```

## Experiment 3: GroupKFold index demonstration

**Purpose.** GroupKFold preserves groups in each split but may not enforce a spatial buffer.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import numpy as np
from sklearn.model_selection import GroupKFold
X=np.arange(12).reshape(-1,1); y=np.arange(12)
groups=np.repeat(np.arange(4),3)
for train,test in GroupKFold(n_splits=4).split(X,y,groups):
    print('Test groups:', np.unique(groups[test]))
```

### How to interpret this experiment

GroupKFold preserves groups in each split but may not enforce a spatial buffer.

- `GroupKFold` — Keeps the same spatial group within one validation fold.
