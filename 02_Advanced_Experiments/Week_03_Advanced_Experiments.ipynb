## Experiment 3: Compute sample standard deviation

**Purpose.** Sample SD uses the n−1 denominator.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import pandas as pd
obs = pd.read_csv(DATA / 'observations_clean.csv')
x = obs['lst_c'].dropna()
print('n=', len(x), 'SD=', x.std(ddof=1))
```

## Experiment 4: Paired correlation

**Purpose.** A paired correlation only uses rows with both relevant variables available.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import pandas as pd
cells = pd.read_csv(DATA / 'cells.csv')
obs = pd.read_csv(DATA / 'observations_clean.csv')
june = cells.merge(obs[obs.date == '2025-06-01'], on='cell_id')
pair = june[['canopy_frac','lst_c']].dropna()
print(pair.corr(method='spearman'))
```

### How to interpret this experiment

A paired correlation only uses rows with both relevant variables available.

- `pd.read_csv` — Reads CSV text into a pandas DataFrame. Inspect head() and column types after loading.
- `merge` — Joins tables on a declared key; validate cardinality to avoid duplicate matches.
