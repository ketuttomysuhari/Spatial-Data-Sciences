## Experiment 1: Compute a review threshold

**Purpose.** A percentile threshold creates a reproducible example review rule.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import pandas as pd
obs=pd.read_csv(DATA/'observations_clean.csv')
june=obs[obs.date=='2025-06-01']
threshold=june['lst_c'].quantile(.75)
print('Q75:',threshold)
```

### How to interpret this experiment

A percentile threshold creates a reproducible example review rule.

## Experiment 2: Join flags to cells

**Purpose.** Join tables with validated keys to avoid misassigning flags.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import pandas as pd
cells=pd.read_csv(DATA/'cells.csv')
obs=pd.read_csv(DATA/'observations_clean.csv')
june=obs[obs.date=='2025-06-01']
a=cells.merge(june,on='cell_id',validate='one_to_one')
a['review']=(a['lst_c']>=a['lst_c'].quantile(.75)).astype(int)
print(a['review'].value_counts())
```

### How to interpret this experiment

Join tables with validated keys to avoid misassigning flags.

- `pd.read_csv` — Reads CSV text into a pandas DataFrame. Inspect head() and column types after loading.
- `merge` — Joins tables on a declared key; validate cardinality to avoid duplicate matches.
- `quantile` — Returns a data percentile for thresholds or distribution summaries.

**Questions:** (1) What does the printed output measure? (2) What happens if one input or parameter changes? (3) How could this method be extended to an independent geographic dataset?
