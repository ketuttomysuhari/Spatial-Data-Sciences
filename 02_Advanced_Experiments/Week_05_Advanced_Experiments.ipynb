## Experiment 1: Mean baseline

**Purpose.** Always compare a fitted regressor with a training-only baseline.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import numpy as np
ytrain=np.array([30.,32.,34.])
ytest=np.array([31.,35.])
pred=np.full(len(ytest), ytrain.mean())
mae=np.mean(np.abs(pred-ytest))
print('Baseline MAE:', mae)
```

## Experiment 2: Ridge with safe preprocessing

**Purpose.** Imputation and scaling belong inside the model pipeline to prevent leakage.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import numpy as np
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
X=np.array([[.1,1],[.4,2],[.6,3],[.8,4]],dtype=float)
y=np.array([31.,32.,34.,36.])
model=make_pipeline(SimpleImputer(),StandardScaler(),Ridge(alpha=1))
model.fit(X[:3],y[:3]); print(model.predict(X[3:]))
```

### How to interpret this experiment

Imputation and scaling belong inside the model pipeline to prevent leakage.

- `SimpleImputer` — Replaces missing predictors with a statistic learned on training data.
- `StandardScaler` — Centers and scales columns using values learned from training data.

**Questions:** (1) What does the printed output measure? (2) What happens if one input or parameter changes? (3) How could this method be extended to an independent geographic dataset?
