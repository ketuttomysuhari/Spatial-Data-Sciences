## Experiment 1: Binary mask overlap

**Purpose.** IoU measures foreground overlap using intersection and union.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import numpy as np
ref=np.array([[1,0,1],[0,1,0]],dtype=bool)
pred=np.array([[1,1,0],[0,1,0]],dtype=bool)
tp=np.count_nonzero(ref & pred)
union=np.count_nonzero(ref | pred)
print('IoU:',tp/union)
```

### How to interpret this experiment

IoU measures foreground overlap using intersection and union.

- `np.count_nonzero` — Counts True Boolean conditions for confusion-matrix or IoU calculations.

## Experiment 3: Threshold selection on train region

**Purpose.** Select a threshold on development data, not on the hold-out region.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import numpy as np
ref=np.array([1,1,0,0],dtype=bool); score=np.array([.9,.6,.5,.2])
def iou(t):
    m=score>=t
    return np.count_nonzero(ref & m)/max(1,np.count_nonzero(ref | m))
print([(t,iou(t)) for t in [.3,.5,.7]])
```

### How to interpret this experiment

Select a threshold on development data, not on the hold-out region.

- `np.count_nonzero` — Counts True Boolean conditions for confusion-matrix or IoU calculations.
