## Experiment 1: Compute evaluation metrics

**Purpose.** MAE, RMSE and bias capture different aspects of error.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import numpy as np
obs=np.array([30.,32.,34.,36.]); pred=np.array([31.,31.,35.,37.])
e=pred-obs
print('MAE',np.abs(e).mean())
print('RMSE',np.sqrt(np.mean(e**2)))
print('Bias',e.mean())
```

### How to interpret this experiment

MAE, RMSE and bias capture different aspects of error.

## Experiment 2: Calculate R squared explicitly

**Purpose.** R² can be negative on an independent test set.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import numpy as np
y=np.array([30.,32.,34.,36.]); p=np.array([31.,31.,35.,37.])
sse=np.sum((p-y)**2); sst=np.sum((y-y.mean())**2)
r2=1-sse/sst
print('R2:',r2)
```

## Experiment 3: Compare model and mean baseline

**Purpose.** A training-mean baseline is a deployable comparator.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import numpy as np
ytrain=np.array([30.,32.,34.]); ytest=np.array([33.,35.])
model=np.array([32.,34.]); baseline=np.full(2,ytrain.mean())
print('Model MAE:',np.abs(model-ytest).mean())
print('Baseline MAE:',np.abs(baseline-ytest).mean())
```

## Experiment 4: Inspect errors by group

**Purpose.** Group-level residual summaries can identify geographic bias.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import pandas as pd
df=pd.DataFrame({'group':['A','A','B','B'],'obs':[30,32,34,36],'pred':[31,31,35,37]})
df['abs_error']=(df['pred']-df['obs']).abs()
print(df.groupby('group')['abs_error'].agg(['count','mean']))
```

### How to interpret this experiment

Group-level residual summaries can identify geographic bias.

- `groupby` — Groups rows before counting or aggregating by sector, date, or other key.
