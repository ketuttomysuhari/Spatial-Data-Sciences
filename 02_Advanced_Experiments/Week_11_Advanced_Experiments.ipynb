## Experiment 2: Calculate confusion counts

**Purpose.** A confusion matrix keeps all four error categories visible.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
from sklearn.metrics import confusion_matrix
truth=[0,1,0,1,1]; flag=[0,0,0,1,1]
cm=confusion_matrix(truth,flag,labels=[0,1])
print(cm)
```

## Experiment 3: Precision and recall safely

**Purpose.** Precision and recall depend on the classifier threshold and case prevalence.

## Experiment 4: Threshold sensitivity

**Purpose.** Changing a threshold trades off missed events against false alarms.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import numpy as np
scores=np.array([.2,.4,.6,.8]); cutoffs=[.3,.5,.7]
for t in cutoffs:
    print(t, int((scores>=t).sum()))
```
