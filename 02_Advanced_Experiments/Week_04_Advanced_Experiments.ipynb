## Experiment 1: Create rook weights for a 3x3 grid

**Purpose.** Rook contiguity links cells that share an edge, not merely a vertex.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import numpy as np
W = np.zeros((9,9), dtype=int)
for i in range(9):
    r, c = divmod(i, 3)
    for dr,dc in [(1,0),(-1,0),(0,1),(0,-1)]:
        nr,nc = r+dr,c+dc
        if 0<=nr<3 and 0<=nc<3: W[i,3*nr+nc]=1
print(W.sum(axis=1))
```

### How to interpret this experiment

Rook contiguity links cells that share an edge, not merely a vertex.

## Experiment 2: Calculate a spatial lag

## Experiment 3: Moran I toy example

**Purpose.** Moran I compares deviations and nearby similarity, not causality.

**Scientific task:** Read the snippet, predict the output, run it, then explain the result and the assumptions.

```python
import numpy as np
z=np.array([-1.,0.,1.])
W=np.array([[0,1,0],[1,0,1],[0,1,0]],dtype=float)
I=(len(z)/W.sum())*(z @ W @ z)/(z @ z)
print('Moran I:', I)
```

### How to interpret this experiment

Moran I compares deviations and nearby similarity, not causality.
