## 3. Main computation

Run this cell after the setup. The block calculates the principal statistics or fits the weekly model. The original course code is retained so results remain aligned with the slide examples.

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,classification_report
cells=pd.read_csv(DATA/'cells.csv');obs=pd.read_csv(DATA/'observations_clean.csv')
may=obs[obs.date=='2025-05-01'][['cell_id','lst_c']].rename(columns={'lst_c':'may_lst'})
june=obs[obs.date=='2025-06-01'][['cell_id','lst_c']]
df=cells.merge(may,on='cell_id').merge(june,on='cell_id').dropna(subset=['may_lst','lst_c'])
df['review']=((df.lst_c-df.may_lst)>=.55).astype(int)
FEATURES=['canopy_frac','impervious_frac','elevation_m','may_lst']
dev=df[df.col<9];east=df[df.col>=9]
model=Pipeline([('imp',SimpleImputer(strategy='median')),('scale',StandardScaler()),('clf',LogisticRegression(max_iter=1000))]);model.fit(dev[FEATURES],dev.review)
score=model.predict_proba(east[FEATURES])[:,1]
flag=(score>=.5).astype(int)
print('Event threshold: June - May ≥ 0.55 °C; score threshold: 0.50')
```

### Reading the main code

**Create labels.** A synthetic May→June temperature-change threshold defines positive review cases. June target values must never leak into predictors.

- `Pipeline` — Combines transformations and estimator, fitting preprocessing only on training rows.
- `SimpleImputer` — Replaces missing predictors with a statistic learned on training data.
- `StandardScaler` — Centers and scales columns using values learned from training data.
- `confusion_matrix` — Tabulates actual versus predicted binary labels.
- `pd.read_csv` — Reads CSV text into a pandas DataFrame. Inspect head() and column types after loading.
- `merge` — Joins tables on a declared key; validate cardinality to avoid duplicate matches.

**Verification questions:** Which variables contain the inputs? What new object does the cell create? What numerical result should be checked before continuing?

## 4. Map, plot, or evaluate the result

The final block turns results into a labelled graphic or output artifact. Keep units in axes/legends, and distinguish model predictions from observations.

```python
cm=confusion_matrix(east.review,flag,labels=[0,1]);print(cm)
print(classification_report(east.review,flag,zero_division=0))
plt.imshow(cm,cmap='Blues');plt.xticks([0,1],['No flag','Flag']);plt.yticks([0,1],['No event','Event']);
for i in range(2):
 for j in range(2):plt.text(j,i,str(cm[i,j]),ha='center',va='center')
plt.title('Held-out human-review queue');plt.show()
```

### Reading the output code

**Train classifier.** A logistic model estimates scores from features; the cutoff creates operational flags.

- `confusion_matrix` — Tabulates actual versus predicted binary labels.
- `plt.` — Creates a figure; title, axes, and legend are part of the analytical interpretation.

**Evaluate decisions.** A confusion matrix shows false negatives and false positives; precision/recall depend on the threshold and class balance.
