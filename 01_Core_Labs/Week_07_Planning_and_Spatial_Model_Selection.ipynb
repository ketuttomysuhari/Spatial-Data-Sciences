## 3. Main computation

Run this cell after the setup. The block calculates the principal statistics or fits the weekly model. The original course code is retained so results remain aligned with the slide examples.

```python
from sklearn.model_selection import GroupKFold,cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.dummy import DummyRegressor
cells=pd.read_csv(DATA/'cells.csv');obs=pd.read_csv(DATA/'observations_clean.csv')
june=cells.merge(obs[obs.date=='2025-06-01'],on='cell_id');dev=june[june.col<9]
X=dev[['canopy_frac','impervious_frac','elevation_m']];y=dev.lst_c
cv=GroupKFold(n_splits=4)
candidates={'Mean':DummyRegressor(),**{f'Ridge {v}':Pipeline([('imp',SimpleImputer(strategy='median')),('scale',StandardScaler()),('reg',Ridge(alpha=v))]) for v in [.1,1,10]}}
scores={name:-cross_val_score(model,X,y,cv=cv,groups=dev.block_id,scoring='neg_mean_absolute_error') for name,model in candidates.items()}
for name,s in scores.items(): print(name,'mean MAE',round(s.mean(),3),'sd',round(s.std(),3))
selected=min(scores,key=lambda k:scores[k].mean());print('Selected:',selected)
```

### Reading the main code

**Development-only tuning.** Keep the eastern geography untouched until Week 08, and use `GroupKFold` on development blocks.

- `GroupKFold` — Keeps the same spatial group within one validation fold.
- `Pipeline` — Combines transformations and estimator, fitting preprocessing only on training rows.
- `SimpleImputer` — Replaces missing predictors with a statistic learned on training data.
- `StandardScaler` — Centers and scales columns using values learned from training data.
- `pd.read_csv` — Reads CSV text into a pandas DataFrame. Inspect head() and column types after loading.
- `merge` — Joins tables on a declared key; validate cardinality to avoid duplicate matches.
- `cross_val_score` — Computes validation scores under a specified cross-validation scheme.
